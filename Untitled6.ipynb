```python
import pandas as pd

# Charger le fichier Excel
file_path = '/content/nina.xlsx'

# Lire toutes les feuilles du fichier Excel pour voir sa structure
excel_file = pd.ExcelFile(file_path)
print("Feuilles disponibles dans le fichier :", excel_file.sheet_names)

# Charger la première feuille par défaut
df = pd.read_excel(file_path, sheet_name=0)

# Afficher les premières lignes
display(df.head())
```

### Définition des variables pour l'Apprentissage Supervisé (X et Y)

* **Cible ($y$)** : `diabetic` (Variable binaire à prédire)
* **Caractéristiques ($X$)** : Les données cliniques et démographiques du patient (âge, genre, IMC, pression artérielle, glucose, antécédents, etc.)

```python
# 1. Définition de la variable cible Y
y = df['diabetic']

# 2. Définition des variables explicatives X (on exclut la cible)
X = df.drop(columns=['diabetic'])

# Affichage des dimensions de X et Y
print("Dimensions des caractéristiques (X) :", X.shape)
print("Dimensions de la cible (y) :", y.shape)

# Aperçu de X et y
print("\n--- Liste des variables explicatives (X) ---")
print(X.columns.tolist())

print("\n--- Distribution de la variable cible (y) ---")
print(y.value_counts(dropna=False))
```

### Étape de Prétraitement des Données

1. Encodage de la cible $y$ (`Yes` $\rightarrow$ 1, `No` $\rightarrow$ 0).
2. Encodage des variables catégorielles dans $X$ (comme `gender`).

```python
# 1. Encoder la variable cible y en numérique (0 et 1)
y_encoded = y.map({'No': 0, 'Yes': 1})

# 2. Encoder les variables catégorielles de X (comme 'gender') avec One-Hot Encoding ou Label Encoding
X_encoded = pd.get_dummies(X, columns=['gender'], drop_first=True)

# Affichage des premières lignes de nos nouvelles matrices prêtes pour le Machine Learning
print("Aperçu des caractéristiques prétraitées (X_encoded) :")
display(X_encoded.head())

print("\nAperçu de la cible encodée (y_encoded) :")
print(y_encoded.head())
```

### Construction et Évaluation du Modèle de Machine Learning (Random Forest)

Nous allons séparer nos données en ensembles d'entraînement et de test, entraîner un classificateur Random Forest, puis analyser ses performances de prédiction.

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score

# 1. Division des données (75% entraînement, 25% test pour l'évaluation finale)
X_train, X_test, y_train, y_test = train_test_split(X_encoded, y_encoded, test_size=0.25, random_state=42, stratify=y_encoded)

# Affichage du nombre final de lignes pour chaque ensemble
print(f"Nombre total de lignes dans le jeu de données : {len(X_encoded)}")
print(f"Nombre de lignes pour l'entraînement (75%) : {len(X_train)}")
print(f"Nombre de lignes pour le test final (25%) : {len(X_test)}\n")

# 2. Initialisation et entraînement du modèle Random Forest
model = RandomForestClassifier(random_state=42, n_estimators=100)
model.fit(X_train, y_train)

# 3. Prédictions sur l'ensemble de test
y_pred = model.predict(X_test)

# 4. Évaluation du modèle
accuracy = accuracy_score(y_test, y_pred)
print(f"Précision globale du modèle (Accuracy) : {accuracy:.2%}\n")
print("--- Rapport de Classification ---")
print(classification_report(y_test, y_pred, target_names=['Non Diabétique (0)', 'Diabétique (1)']))
```

### Évaluation Statistique Approfondie (Matrice de Confusion)

```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix

# Calcul de la matrice de confusion
cm = confusion_matrix(y_test, y_pred)

# Affichage visuel sous forme de carte de chaleur
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Non Diabétique (0)', 'Diabétique (1)'],
            yticklabels=['Non Diabétique (0)', 'Diabétique (1)'])
plt.ylabel('Classe Réelle')
plt.xlabel('Classe Prédite')
plt.title('Matrice de Confusion du Modèle')
plt.show()

# Calcul détaillé des taux statistiques
tn, fp, fn, tp = cm.ravel()
sensitivity_recall = tp / (tp + fn)
specificity = tn / (tn + fp)
precision = tp / (tp + fp)

print("--- Analyse Métrique / Normes Statistiques ---")
print(f"Vrais Négatifs (Sains correctement détectés) : {tn}")
print(f"Faux Positifs (Sains classés malades à tort) : {fp}")
print(f"Faux Négatifs (Malades non détectés) : {fn}  <-- Indicateur critique à réduire !")
print(f"Vrais Positifs (Malades correctement détectés) : {tp}\n")
print(f"Sensibilité (Rappel / True Positive Rate) : {sensitivity_recall:.2%} (Capacité à détecter les malades)")
print(f"Spécificité (True Negative Rate) : {specificity:.2%} (Capacité à détecter les bien portants)")
print(f"Valeur Prédictive Positive (Précision) : {precision:.2%}")
```
